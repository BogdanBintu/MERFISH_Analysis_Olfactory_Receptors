# This Notebook given an example on how to plot the number of ONS per section and the average single-cell expression of receptors. (Figure 1 in the manuscript)

```python
import pandas as pd
import numpy as np
import matplotlib.pylab as plt
```

```python
import pandas as pd
T1 = r'C:\Users\Bogdan\Dropbox\ORFigures\Figures2024\Tables\Table S2.xlsx'
df = pd.read_excel(T1, sheet_name=0,skiprows=1)
# Display the first few rows of the DataFrame
print(df.head())
dic_cts_female = dict(zip(df['Receptor Name'],df['Median number of OSNs per section']))

df = pd.read_excel(T1, sheet_name=1,skiprows=1)
# Display the first few rows of the DataFrame
print(df.head())
dic_cts_male = dict(zip(df['Receptor Name'],df['Median number of OSNs per section']))


df = pd.read_excel(T1, sheet_name=2,skiprows=1)
# Display the first few rows of the DataFrame
print(df.head())
dic_expr_female = dict(zip(df['Receptor Name'],df['Median single-cell expression level']))

df = pd.read_excel(T1, sheet_name=3,skiprows=1)
# Display the first few rows of the DataFrame
print(df.head())
dic_expr_male = dict(zip(df['Receptor Name'],df['Median single-cell expression level']))
```

```python
import matplotlib as mpl
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['font.size'] = 15
ORs = list(dic_cts_female.keys())
X1 = list(dic_cts_female.values())
X2 = list(dic_cts_male.values())

keep = ~((X1==0)|(X2==0)|np.isnan(X1)|np.isnan(X2))
rho = np.round(np.corrcoef(np.log(X1)[keep],np.log(X2)[keep]),1)[0,1]

plt.loglog(X1,X2,'o',c='gray')
plt.loglog([np.nanmin(X1),np.nanmax(X1)],[np.nanmin(X2),np.nanmax(X2)],'r-',lw=3,label='ρ='+str(rho))
plt.xlabel('Average number of OSNs per section (females)')
plt.ylabel('Average number of OSNs per section (males)')
plt.legend();
```

```python
import matplotlib as mpl
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['font.size'] = 15
ORs = list(dic_expr_female.keys())
X1 = list(dic_expr_female.values())
X2 = list(dic_expr_male.values())

keep = ~(np.isnan(X1)|np.isnan(X2))
rho = np.round(np.corrcoef(np.array(X1)[keep],np.array(X2)[keep]),1)[0,1]

plt.plot(X1,X2,'o',c='gray')
plt.plot([np.nanmin(X1),np.nanmax(X1)],[np.nanmin(X2),np.nanmax(X2)],'r-',lw=3,label = 'ρ='+str(rho))
plt.xlabel('Average receptor expression per cell (females)')
plt.ylabel('Average receptor expression per cell (males)');
plt.legend()
#plt.xlim([1.5,6])
#plt.ylim([2,6])
```
